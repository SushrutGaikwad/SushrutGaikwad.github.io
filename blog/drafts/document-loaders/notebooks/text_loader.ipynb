```python
%%capture
from langchain_community.document_loaders import TextLoader
from pathlib import Path
```

We will work with the `transformers.txt` file.

```python
# Define the path

file_path = Path("transformers.txt")  # path to the `transformers.txt` file
file_path.exists()  # check if the file exists on the defined path
```

```python
# Define the text loader

loader = TextLoader(file_path=file_path, encoding="utf-8")
loader
```

```python
# Load the documents

documents = loader.load()  # gives a list of `Document` objects
type(documents)
```

```python
# Number of documents

len(documents)
```

Our list will only contain a single `Document` object. But if we were using a PDF loader, it would have given us a `Document` object per page in this list.

```python
print(type(documents[0]))
```

```python
documents
```

We can clearly see the two attributes `metadata` which is a dictionary that currently contains the source information, and `page_content` which contains the actual text present in the file.

```python
extracted_doc = documents[0]
```

```python
# page_content

print(extracted_doc.page_content)
```

```python
# metadata

print(extracted_doc.metadata)
```
